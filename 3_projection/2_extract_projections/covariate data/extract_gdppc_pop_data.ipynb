```python
import xarray as xr
import os
import pandas as pd

def open_dscim_econ_vars(ssp_in, model_in, year_list, econ_vars, 
path_in='/project/cil/gcp/integration/float32/dscim_input_data/econvars/zarrs/integration-econ-bc39.zarr'):

    econ_var_data = xr.open_zarr(path_in)
    
    econ_var_out = (econ_var_data[econ_vars].sel(ssp = ssp_in, model = model_in, year = year_list).to_dataframe()) 
                    
    econ_var_out.reset_index(inplace = True)
                    
    return econ_var_out
```
