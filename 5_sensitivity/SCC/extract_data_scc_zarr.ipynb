```python
import xarray as xr
import pandas as pd
import os

# 启用 Zarr v3 API
os.environ['ZARR_V3_EXPERIMENTAL_API'] = '1'

# 定义函数（添加 rcp 参数）
def open_dscim_econ_vars(ssp_in, model_in, rcp_in, year_list, econ_vars, path_in):
    """
    从 zarr 文件中提取经济变量数据
    
    Parameters:
    -----------
    ssp_in : str
        SSP scenario (e.g., "SSP3")
    model_in : str
        Model name (e.g., "IIASA GDP")
    rcp_in : str
        RCP scenario (e.g., "rcp85")
    year_list : list
        List of years to extract
    econ_vars : list
        List of economic variables to extract
    path_in : str
        Path to zarr file
    
    Returns:
    --------
    DataFrame with extracted data
    """
    # 打开 zarr 文件
    econ_var_data = xr.open_zarr(path_in, zarr_version=3)
    
    # 选择变量和维度（添加 rcp）
    econ_var_out = (econ_var_data[econ_vars]
                    .sel(ssp=ssp_in, model=model_in, rcp=rcp_in, year=year_list)
                    .to_dataframe())
    
    # 重置索引
    econ_var_out.reset_index(inplace=True)
    
    return econ_var_out

# 使用函数（添加 rcp_in 参数）
df = open_dscim_econ_vars(
    ssp_in="SSP3",
    model_in="IIASA GDP",
    rcp_in="rcp85",  # RCP 8.5
    year_list=[2099],
    econ_vars=["cc"],  # 或 ["anomaly", "damages"]
    path_in="/project/cil/home_dirs/scadavidsanchez/repos/dscim-labor-2025/pulse_2025/reduced_damages_labor_old_mc/labor/risk_aversion_cc_eta2.0.zarr"
)

df.head()
```

```python
import os
import xarray as xr
import zarr

# 启用 Zarr v3 实验性 API
os.environ['ZARR_V3_EXPERIMENTAL_API'] = '1'

path = "/project/cil/home_dirs/scadavidsanchez/repos/dscim-labor-2025/results-2025-pulse-bu/reduced_damages_labor_old_mc/labor/risk_aversion_cc_eta2.0.zarr"

# 方法1：使用 xarray 读取 zarr v3
try:
    ds = xr.open_zarr(path, consolidated=False, zarr_version=3)
    print("✓ 成功打开！")
    print("\nDataset 信息:")
    print(ds)
    print("\n变量列表:")
    print(list(ds.data_vars))
    print("\n维度:")
    print(ds.dims)
except Exception as e:
    print(f"✗ xarray 打开失败: {e}")

# 方法2：使用 zarr 直接打开
try:
    root = zarr.open_group(path, mode='r')
    print("\n✓ Zarr 打开成功！")
    print("\nRoot 结构:")
    print(root.tree())
    print("\nRoot keys:")
    print(list(root.keys()) if hasattr(root, 'keys') else 'No keys')
except Exception as e:
    print(f"\n✗ Zarr 打开失败: {e}")
```
